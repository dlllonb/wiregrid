```python
def load_grasp_grd(filepath):
    """
    Load a GRASP .grd file with complex field components.

    Returns:
        x, y       : 1D coordinate arrays
        F1, F2     : complex 2D arrays (ny, nx)
    """
    with open(filepath, 'r') as f:
        lines = f.readlines()

    # --- Find the line with grid size (nx ny 0)
    grid_idx = None
    for i, line in enumerate(lines):
        parts = line.strip().split()
        if len(parts) == 3:
            try:
                nx = int(parts[0])
                ny = int(parts[1])
                grid_idx = i
                break
            except:
                continue

    if grid_idx is None:
        raise ValueError("Could not find grid size line.")

    # --- Data starts immediately after grid line
    data = np.loadtxt(lines[grid_idx + 1:])

    # --- Extract fields
    F1 = data[:, 0] + 1j * data[:, 1]
    F2 = data[:, 2] + 1j * data[:, 3]

    # --- Reshape
    F1 = F1.reshape((ny, nx))
    F2 = F2.reshape((ny, nx))

    # --- Build coordinate arrays from bounds line
    # Find bounds line (xmin ymin xmax ymax)
    for i in range(grid_idx):
        parts = lines[i].strip().split()
        if len(parts) == 4:
            try:
                xmin, ymin, xmax, ymax = map(float, parts)
            except:
                continue

    x = np.linspace(xmin, xmax, nx)
    y = np.linspace(ymin, ymax, ny)

    return x, y, F1, F2


def plot_beam(x, y, F1, F2, title="Beam"):
    """
    Quick visualization of total intensity and components.
    """
    I = np.abs(F1)**2 + np.abs(F2)**2

    fig, axs = plt.subplots(1, 3, figsize=(15, 4))

    im0 = axs[0].imshow(np.abs(F1), extent=[x.min(), x.max(), y.min(), y.max()], origin='lower')
    axs[0].set_title("|F1|")
    plt.colorbar(im0, ax=axs[0])

    im1 = axs[1].imshow(np.abs(F2), extent=[x.min(), x.max(), y.min(), y.max()], origin='lower')
    axs[1].set_title("|F2|")
    plt.colorbar(im1, ax=axs[1])

    im2 = axs[2].imshow(I, extent=[x.min(), x.max(), y.min(), y.max()], origin='lower')
    axs[2].set_title("Total Intensity")
    plt.colorbar(im2, ax=axs[2])

    for ax in axs:
        ax.set_xlabel("x")
        ax.set_ylabel("y")

    plt.suptitle(title)
    plt.tight_layout()
    plt.show()
```

```python
# Load both
x_off, y_off, F1_off, F2_off = load_grasp_grd(data_dir + "azel_1deg_offy.grd")
x_on,  y_on,  F1_on,  F2_on  = load_grasp_grd(data_dir + "azel_1deg_qps6000w_y.grd")

# Plot
plot_beam(x_off, y_off, F1_off, F2_off, title="Wire OFF")
plot_beam(x_on,  y_on,  F1_on,  F2_on,  title="Wire ON")
```

```python
def compute_stokes(F1, F2):
    """
    Compute Stokes parameters from two complex field components
    in a linear polarization basis.

    Parameters
    ----------
    F1, F2 : 2D complex ndarray
        Complex field components on the same grid.

    Returns
    -------
    I, Q, U, V : 2D float ndarray
        Stokes parameter maps.
    """
    E1E1 = np.abs(F1)**2
    E2E2 = np.abs(F2)**2
    cross = F1 * np.conj(F2)

    I = E1E1 + E2E2
    Q = E1E1 - E2E2
    U = 2.0 * np.real(cross)
    V = -2.0 * np.imag(cross)

    return I, Q, U, V


def plot_stokes(x, y, I, Q, U, V=None, title="Stokes Maps", use_symlog=False):
    """
    Plot Stokes parameter maps for initial inspection.

    Parameters
    ----------
    x, y : 1D ndarray
        Coordinate axes.
    I, Q, U : 2D ndarray
        Stokes maps.
    V : 2D ndarray or None
        Optional V map.
    title : str
        Overall figure title.
    use_symlog : bool
        If True, show Q/U/V with symmetric log scaling by plotting
        sign(data)*log10(1+abs(data)). Useful if dynamic range is large.
    """
    maps = [I, Q, U] if V is None else [I, Q, U, V]
    names = ["I", "Q", "U"] if V is None else ["I", "Q", "U", "V"]

    ncols = len(maps)
    fig, axs = plt.subplots(1, ncols, figsize=(5 * ncols, 4))

    if ncols == 1:
        axs = [axs]

    extent = [x.min(), x.max(), y.min(), y.max()]

    for ax, data, name in zip(axs, maps, names):
        display_data = data.copy()

        if use_symlog and name != "I":
            display_data = np.sign(data) * np.log10(1.0 + np.abs(data))

        im = ax.imshow(display_data, extent=extent, origin="lower", aspect="equal")
        ax.set_title(name if not use_symlog or name == "I" else f"{name}")
        ax.set_xlabel("x")
        ax.set_ylabel("y")
        plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

    fig.suptitle(title)
    plt.tight_layout()
    plt.show()


def summarize_stokes(I, Q, U, V=None, label="dataset"):
    """
    Print a few quick summary stats for sanity checking.
    """
    print(f"\n--- {label} ---")
    print(f"I: min={I.min():.6e}, max={I.max():.6e}")
    print(f"Q: min={Q.min():.6e}, max={Q.max():.6e}")
    print(f"U: min={U.min():.6e}, max={U.max():.6e}")
    if V is not None:
        print(f"V: min={V.min():.6e}, max={V.max():.6e}")

    frac_pol = np.sqrt(Q**2 + U**2) / np.maximum(I, 1e-30)
    print(f"sqrt(Q^2+U^2)/I: min={frac_pol.min():.6e}, max={frac_pol.max():.6e}")
```

```python
# Compute Stokes for wire OFF
I_off, Q_off, U_off, V_off = compute_stokes(F1_off, F2_off)

# Compute Stokes for wire ON
I_on, Q_on, U_on, V_on = compute_stokes(F1_on, F2_on)

# Quick summaries
summarize_stokes(I_off, Q_off, U_off, V_off, label="Wire OFF")
summarize_stokes(I_on, Q_on, U_on, V_on, label="Wire ON")

# Initial plots
plot_stokes(x_off, y_off, I_off, Q_off, U_off, V_off, title="Wire OFF Stokes", use_symlog=True)
plot_stokes(x_on, y_on, I_on, Q_on, U_on, V_on, title="Wire ON Stokes", use_symlog=True)
```

```python

def compute_polarization_angle(Q, U, I=None, pol_fraction_floor=None, I_floor=None, degrees=True):
    """
    Compute linear polarization angle from Stokes Q and U.

    Uses the correct convention:
        psi = 0.5 * atan2(U, Q)

    Parameters
    ----------
    Q, U : 2D ndarray
        Stokes Q and U maps.
    I : 2D ndarray or None
        Optional Stokes I map, used for masking if desired.
    pol_fraction_floor : float or None
        If set, mask pixels where sqrt(Q^2 + U^2)/I < pol_fraction_floor.
        Requires I.
    I_floor : float or None
        If set, mask pixels where I < I_floor.
    degrees : bool
        If True, return angle in degrees. Otherwise radians.

    Returns
    -------
    psi : 2D ndarray
        Polarization angle map, with masked/invalid pixels set to NaN.
    """
    psi = 0.5 * np.arctan2(U, Q)

    mask = np.zeros(Q.shape, dtype=bool)

    if I_floor is not None:
        if I is None:
            raise ValueError("I must be provided if using I_floor.")
        mask |= (I < I_floor)

    if pol_fraction_floor is not None:
        if I is None:
            raise ValueError("I must be provided if using pol_fraction_floor.")
        p_lin = np.sqrt(Q**2 + U**2)
        frac_pol = p_lin / np.maximum(I, 1e-30)
        mask |= (frac_pol < pol_fraction_floor)

    psi = np.where(mask, np.nan, psi)

    if degrees:
        psi = np.degrees(psi)

    return psi


def plot_polarization_angle(x, y, psi, title="Polarization Angle", cmap="twilight"):
    """
    Plot polarization angle map.
    """
    fig, ax = plt.subplots(figsize=(6, 5))

    im = ax.imshow(
        psi,
        extent=[x.min(), x.max(), y.min(), y.max()],
        origin="lower",
        aspect="equal",
        cmap=cmap
    )

    ax.set_title(title)
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    plt.colorbar(im, ax=ax, label="Angle [deg]")
    plt.tight_layout()
    plt.show()


def summarize_polarization_angle(psi, label="dataset"):
    """
    Print quick summary stats for polarization angle map, ignoring NaNs.
    """
    valid = np.isfinite(psi)
    if not np.any(valid):
        print(f"\n--- {label} ---")
        print("No valid pixels in polarization angle map.")
        return

    vals = psi[valid]
    print(f"\n--- {label} polarization angle ---")
    print(f"valid pixels: {valid.sum()} / {valid.size}")
    print(f"min = {np.nanmin(vals):.6f} deg")
    print(f"max = {np.nanmax(vals):.6f} deg")
    print(f"mean = {np.nanmean(vals):.6f} deg")
    print(f"std = {np.nanstd(vals):.6f} deg")

```

```python
def wrap_linear_pol_angle_deg(angle_deg):
    """
    Wrap a linear polarization angle into [-90, 90) degrees.
    """
    return ((angle_deg + 90.0) % 180.0) - 90.0


def polarization_angle_difference_deg(psi_on_deg, psi_off_deg):
    """
    Compute the difference between two linear polarization angle maps
    in degrees, correctly handling the 180-degree ambiguity.

    Returns delta_psi in [-90, 90) degrees.
    """
    # Work in doubled-angle space to respect linear polarization symmetry
    two_on = np.radians(2.0 * psi_on_deg)
    two_off = np.radians(2.0 * psi_off_deg)

    # Wrapped difference in doubled-angle space
    delta_two = np.arctan2(np.sin(two_on - two_off), np.cos(two_on - two_off))

    # Convert back to ordinary linear polarization angle difference
    delta_psi_deg = np.degrees(0.5 * delta_two)

    return delta_psi_deg


def plot_angle_difference(x, y, delta_psi_deg, title=r"$\psi_{\rm on} - \psi_{\rm off}$", cmap="coolwarm",
                          vmin=None, vmax=None):
    """
    Plot polarization angle difference map.
    """
    fig, ax = plt.subplots(figsize=(6, 5))

    if vmin is None or vmax is None:
        finite = np.isfinite(delta_psi_deg)
        if np.any(finite):
            lim = np.nanmax(np.abs(delta_psi_deg[finite]))
            vmin = -lim
            vmax = lim
        else:
            vmin, vmax = -1, 1

    im = ax.imshow(
        delta_psi_deg,
        extent=[x.min(), x.max(), y.min(), y.max()],
        origin="lower",
        aspect="equal",
        cmap=cmap,
        vmin=vmin,
        vmax=vmax
    )

    ax.set_title(title)
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    plt.colorbar(im, ax=ax, label=r"$\Delta \psi$ [deg]")
    plt.tight_layout()
    plt.show()


def summarize_angle_difference(delta_psi_deg, label="delta psi"):
    """
    Print summary statistics for an angle-difference map, ignoring NaNs.
    """
    finite = np.isfinite(delta_psi_deg)
    if not np.any(finite):
        print(f"\n--- {label} ---")
        print("No valid pixels.")
        return

    vals = delta_psi_deg[finite]
    print(f"\n--- {label} ---")
    print(f"valid pixels: {finite.sum()} / {finite.size}")
    print(f"min  = {np.nanmin(vals):.6f} deg")
    print(f"max  = {np.nanmax(vals):.6f} deg")
    print(f"mean = {np.nanmean(vals):.6f} deg")
    print(f"std  = {np.nanstd(vals):.6f} deg")
```

```python
def plot_angle_difference_2(
    x,
    y,
    delta_psi_deg,
    title=r"$\psi_{\rm on} - \psi_{\rm off}$",
    cmap="coolwarm",
    vmin=None,
    vmax=None,
    colorbar_limits=None,
):
    """
    Plot polarization angle difference map.

    Parameters
    ----------
    colorbar_limits : tuple, optional
        (lower, upper) limits for the colorbar. If provided, these override
        vmin and vmax. If None, the color scale is determined from vmin/vmax,
        or defaults to a symmetric range about zero.
    """
    fig, ax = plt.subplots(figsize=(6, 5))

    if colorbar_limits is not None:
        vmin, vmax = colorbar_limits
    elif vmin is None or vmax is None:
        finite = np.isfinite(delta_psi_deg)
        if np.any(finite):
            lim = np.nanmax(np.abs(delta_psi_deg[finite]))
            vmin = -lim
            vmax = lim
        else:
            vmin, vmax = -1, 1

    im = ax.imshow(
        delta_psi_deg,
        extent=[x.min(), x.max(), y.min(), y.max()],
        origin="lower",
        aspect="equal",
        cmap=cmap,
        vmin=vmin,
        vmax=vmax,
    )

    ax.set_title(title)
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    plt.colorbar(im, ax=ax, label=r"$\Delta \psi$ [deg]")
    plt.tight_layout()
    plt.show()
```

```python
def normalize_for_display(weight, stretch="sqrt", floor=0.0):
    """
    Normalize a nonnegative weight map to [0, 1] for display.

    Parameters
    ----------
    weight : 2D ndarray
        Beam/intensity-like quantity.
    stretch : str
        One of: "linear", "sqrt", "log"
    floor : float
        Optional fractional floor relative to max(weight) to clip low values.

    Returns
    -------
    w : 2D ndarray
        Normalized display weight in [0, 1].
    """
    w = np.array(weight, dtype=float)
    w = np.maximum(w, 0.0)

    wmax = np.nanmax(w)
    if not np.isfinite(wmax) or wmax <= 0:
        return np.zeros_like(w)

    if floor > 0:
        w = np.where(w < floor * wmax, 0.0, w)

    w = w / wmax

    if stretch == "linear":
        pass
    elif stretch == "sqrt":
        w = np.sqrt(w)
    elif stretch == "log":
        # gentle log stretch mapped back to [0,1]
        w = np.log10(1 + 999 * w) / np.log10(1000)
    else:
        raise ValueError("stretch must be 'linear', 'sqrt', or 'log'")

    return np.clip(w, 0.0, 1.0)


def plot_angle_difference_weighted_alpha(
    x, y, delta_psi_deg, weight,
    title=r"$\psi_{\rm on} - \psi_{\rm off}$ (alpha-weighted by beam)",
    cmap="coolwarm",
    stretch="sqrt",
    floor=0.0,
    alpha_min=0.03,
    vmin=None,
    vmax=None
):
    """
    Plot delta_psi with transparency set by beam intensity.

    High-intensity regions appear opaque; low-intensity regions fade out.

    Parameters
    ----------
    x, y : 1D ndarray
        Coordinate axes.
    delta_psi_deg : 2D ndarray
        Polarization angle difference map in degrees.
    weight : 2D ndarray
        Beam intensity map used for weighting (e.g. I_off, I_on, or average).
    stretch : str
        "linear", "sqrt", or "log" stretch for the displayed weight.
    floor : float
        Fraction of peak weight below which alpha is set close to zero.
    alpha_min : float
        Minimum alpha for nonzero-weight pixels, so faint structure is still visible.
    vmin, vmax : float or None
        Color range for delta_psi. If None, use symmetric auto-scaling.
    """
    display_weight = normalize_for_display(weight, stretch=stretch, floor=floor)

    # Turn normalized display weight into alpha map
    alpha = alpha_min + (1.0 - alpha_min) * display_weight
    alpha = np.where(display_weight > 0, alpha, 0.0)

    finite = np.isfinite(delta_psi_deg)
    if vmin is None or vmax is None:
        if np.any(finite):
            lim = np.nanmax(np.abs(delta_psi_deg[finite]))
            vmin, vmax = -lim, lim
        else:
            vmin, vmax = -1.0, 1.0

    fig, ax = plt.subplots(figsize=(6, 5))
    im = ax.imshow(
        delta_psi_deg,
        extent=[x.min(), x.max(), y.min(), y.max()],
        origin="lower",
        aspect="equal",
        cmap=cmap,
        vmin=vmin,
        vmax=vmax,
        alpha=alpha
    )

    ax.set_title(title)
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    plt.colorbar(im, ax=ax, label=r"$\Delta \psi$ [deg]")
    plt.tight_layout()
    plt.show()


def plot_angle_difference_weighted_product(
    x, y, delta_psi_deg, weight,
    title=r"Beam-weighted $\Delta \psi$",
    cmap="coolwarm",
    stretch="sqrt",
    floor=0.0,
    vmin=None,
    vmax=None
):
    """
    Plot delta_psi multiplied by a normalized beam weight.

    This is not the raw angle difference anymore; it is a visualization of
    'angle difference importance' under the chosen weighting.
    """
    display_weight = normalize_for_display(weight, stretch=stretch, floor=floor)
    weighted_map = delta_psi_deg * display_weight

    finite = np.isfinite(weighted_map)
    if vmin is None or vmax is None:
        if np.any(finite):
            lim = np.nanmax(np.abs(weighted_map[finite]))
            vmin, vmax = -lim, lim
        else:
            vmin, vmax = -1.0, 1.0

    fig, ax = plt.subplots(figsize=(6, 5))
    im = ax.imshow(
        weighted_map,
        extent=[x.min(), x.max(), y.min(), y.max()],
        origin="lower",
        aspect="equal",
        cmap=cmap,
        vmin=vmin,
        vmax=vmax
    )

    ax.set_title(title)
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    plt.colorbar(im, ax=ax, label=r"weighted $\Delta \psi$ [deg]")
    plt.tight_layout()
    plt.show()
```

```python
def summarize_delta_psi(delta_psi_deg, weight=None, label="delta psi"):
    """
    Print summary statistics for a polarization angle difference map.

    Parameters
    ----------
    delta_psi_deg : 2D ndarray
        Angle difference map in degrees.
    weight : 2D ndarray or None
        Optional nonnegative weight map (e.g. average beam intensity).
        If provided, weighted statistics are also reported.
    label : str
        Label for printing.
    """
    finite = np.isfinite(delta_psi_deg)
    if not np.any(finite):
        print(f"\n--- {label} ---")
        print("No finite pixels found.")
        return

    vals = delta_psi_deg[finite]

    print(f"\n--- {label} ---")
    print(f"valid pixels: {finite.sum()} / {finite.size}")
    print(f"min                = {np.nanmin(vals):.6e} deg")
    print(f"max                = {np.nanmax(vals):.6e} deg")
    print(f"max |delta psi|    = {np.nanmax(np.abs(vals)):.6e} deg")
    print(f"mean               = {np.nanmean(vals):.6e} deg")
    print(f"std                = {np.nanstd(vals):.6e} deg")
    print(f"RMS                = {np.sqrt(np.nanmean(vals**2)):.6e} deg")
    print(f"median             = {np.nanmedian(vals):.6e} deg")

    # Useful thresholds relative to SO-style 0.1 deg requirement
    for thresh in [1e-3, 1e-2, 1e-1]:
        frac = np.mean(np.abs(vals) > thresh)
        print(f"fraction with |delta psi| > {thresh:.3e} deg = {frac:.6f}")

    if weight is not None:
        w = np.array(weight, dtype=float)
        w = np.where(finite, w, 0.0)
        w = np.where(np.isfinite(w) & (w > 0), w, 0.0)

        wsum = np.sum(w)
        if wsum > 0:
            wmean = np.sum(w * delta_psi_deg) / wsum
            wrms = np.sqrt(np.sum(w * delta_psi_deg**2) / wsum)
            wabsmean = np.sum(w * np.abs(delta_psi_deg)) / wsum

            print("\nWeighted by supplied beam/intensity map:")
            print(f"sum of weights      = {wsum:.6e}")
            print(f"weighted mean       = {wmean:.6e} deg")
            print(f"weighted mean |.|   = {wabsmean:.6e} deg")
            print(f"weighted RMS        = {wrms:.6e} deg")
        else:
            print("\nWeighted stats skipped: no positive finite weights.")


def summarize_delta_psi_in_region(x, y, delta_psi_deg, r_max=None, xlim=None, ylim=None,
                                  weight=None, label="delta psi region"):
    """
    Summarize delta_psi inside a chosen region.

    You can select either:
      - circular region via r_max
      - rectangular region via xlim, ylim

    Parameters
    ----------
    x, y : 1D ndarray
        Coordinate axes.
    delta_psi_deg : 2D ndarray
        Angle difference map.
    r_max : float or None
        Radius for circular cut centered at (0,0).
    xlim, ylim : tuple or None
        Rectangular limits like (-0.05, 0.05).
    weight : 2D ndarray or None
        Optional weights.
    """
    X, Y = np.meshgrid(x, y)

    mask = np.isfinite(delta_psi_deg)

    if r_max is not None:
        mask &= (X**2 + Y**2 <= r_max**2)

    if xlim is not None:
        mask &= (X >= xlim[0]) & (X <= xlim[1])

    if ylim is not None:
        mask &= (Y >= ylim[0]) & (Y <= ylim[1])

    if not np.any(mask):
        print(f"\n--- {label} ---")
        print("No pixels in selected region.")
        return

    vals = delta_psi_deg[mask]

    print(f"\n--- {label} ---")
    print(f"pixels in region    = {mask.sum()}")
    print(f"min                 = {np.nanmin(vals):.6e} deg")
    print(f"max                 = {np.nanmax(vals):.6e} deg")
    print(f"max |delta psi|     = {np.nanmax(np.abs(vals)):.6e} deg")
    print(f"mean                = {np.nanmean(vals):.6e} deg")
    print(f"std                 = {np.nanstd(vals):.6e} deg")
    print(f"RMS                 = {np.sqrt(np.nanmean(vals**2)):.6e} deg")

    if weight is not None:
        w = np.array(weight, dtype=float)
        w = np.where(mask, w, 0.0)
        w = np.where(np.isfinite(w) & (w > 0), w, 0.0)
        wsum = np.sum(w)

        if wsum > 0:
            wmean = np.sum(w * delta_psi_deg) / wsum
            wrms = np.sqrt(np.sum(w * delta_psi_deg**2) / wsum)
            wabsmean = np.sum(w * np.abs(delta_psi_deg)) / wsum

            print("weighted mean        = {:.6e} deg".format(wmean))
            print("weighted mean |.|    = {:.6e} deg".format(wabsmean))
            print("weighted RMS         = {:.6e} deg".format(wrms))
```

```python
# Average beam intensity as a weight map
I_avg = 0.5 * (I_on + I_off)

# Full-map summary
summarize_delta_psi(delta_psi, weight=I_avg, label="Wire ON - OFF delta psi")

# Example: summarize central region within radius 0.05 deg
summarize_delta_psi_in_region(
    x_on, y_on, delta_psi,
    r_max=0.05,
    weight=I_avg,
    label="Wire ON - OFF delta psi inside r < 0.05 deg"
)

# Example: summarize a tighter core if desired
summarize_delta_psi_in_region(
    x_on, y_on, delta_psi,
    r_max=0.02,
    weight=I_avg,
    label="Wire ON - OFF delta psi inside r < 0.02 deg"
)
```

```python
# ------------------------------------------------------------
# Integrated delta-psi inside a beam dB threshold
# Default: integrate only where reference beam is above -20 dB
# ------------------------------------------------------------

def integrate_delta_psi_db_region(
    delta_psi_deg,
    I_ref,
    db_min=-20.0,
    weight=None,
    label="delta psi",
):
    """
    Integrate/summarize delta_psi over pixels where I_ref is above db_min
    relative to max(I_ref).

    Parameters
    ----------
    delta_psi_deg : 2D ndarray
        Polarization angle difference map in degrees.
        Should already be wrapped to [-90, 90) deg.

    I_ref : 2D ndarray
        Reference intensity map used to define the dB mask.
        Usually I_off, I_on, or I_avg.

    db_min : float
        Include pixels with I_ref_dB >= db_min.
        Example: db_min=-20 keeps pixels above -20 dB from peak.

    weight : 2D ndarray or None
        Optional weight map for the integral/statistics.
        If None, all selected pixels are weighted equally.
        A good default choice is I_ref or I_avg.

    label : str
        Printed label.
    """
    delta = np.asarray(delta_psi_deg, dtype=float)
    I_ref = np.asarray(I_ref, dtype=float)

    # dB relative to peak of reference intensity
    I_peak = np.nanmax(I_ref)
    I_db = 10.0 * np.log10(np.maximum(I_ref, 1e-300) / I_peak)

    # Region mask
    mask = np.isfinite(delta) & np.isfinite(I_db) & (I_db >= db_min)

    if weight is None:
        w = np.ones_like(delta, dtype=float)
        weight_label = "unweighted"
    else:
        w = np.asarray(weight, dtype=float)
        mask &= np.isfinite(w) & (w >= 0)
        weight_label = "weighted"

    if not np.any(mask):
        print(f"\n--- {label}: I >= {db_min:.1f} dB ---")
        print("No valid pixels in selected region.")
        return None

    d = delta[mask]
    ww = w[mask]

    # Avoid divide-by-zero
    wsum = np.sum(ww)
    if wsum <= 0:
        print(f"\n--- {label}: I >= {db_min:.1f} dB ---")
        print("Selected pixels have zero total weight.")
        return None

    mean = np.sum(ww * d) / wsum
    mean_abs = np.sum(ww * np.abs(d)) / wsum
    rms = np.sqrt(np.sum(ww * d**2) / wsum)

    # Optional: this is useful for seeing how much area/power is included
    total_finite_pixels = np.count_nonzero(np.isfinite(delta))
    selected_pixels = np.count_nonzero(mask)
    frac_pixels = selected_pixels / total_finite_pixels

    if weight is not None:
        total_weight = np.nansum(w[np.isfinite(w) & (w >= 0)])
        frac_weight = wsum / total_weight if total_weight > 0 else np.nan
    else:
        total_weight = total_finite_pixels
        frac_weight = frac_pixels

    print(f"\n--- {label}: I >= {db_min:.1f} dB ({weight_label}) ---")
    print(f"selected pixels       = {selected_pixels} / {total_finite_pixels}")
    print(f"fraction of pixels    = {frac_pixels:.6e}")
    print(f"sum of weights        = {wsum:.6e}")
    print(f"fraction of weights   = {frac_weight:.6e}")
    print(f"mean delta psi        = {mean:.12e} deg")
    print(f"mean |delta psi|      = {mean_abs:.12e} deg")
    print(f"RMS delta psi         = {rms:.12e} deg")
    print(f"min delta psi         = {np.nanmin(d):.12e} deg")
    print(f"max delta psi         = {np.nanmax(d):.12e} deg")
    print(f"max |delta psi|       = {np.nanmax(np.abs(d)):.12e} deg")

    return {
        "db_min": db_min,
        "weighted": weight is not None,
        "selected_pixels": selected_pixels,
        "fraction_pixels": frac_pixels,
        "sum_weights": wsum,
        "fraction_weights": frac_weight,
        "mean_deg": mean,
        "mean_abs_deg": mean_abs,
        "rms_deg": rms,
        "min_deg": np.nanmin(d),
        "max_deg": np.nanmax(d),
        "max_abs_deg": np.nanmax(np.abs(d)),
        "mask": mask,
        "I_db": I_db,
    }


# Choose reference/weight maps.
# I_avg is usually a good neutral choice for comparing ON vs OFF.
I_avg = 0.5 * (I_on + I_off)

# Main default: beam-weighted integral above -20 dB
result_weighted_20db = integrate_delta_psi_db_region(
    delta_psi,
    I_ref=I_avg,
    db_min=-20.0,
    weight=I_avg,
    label="Wire ON - OFF delta psi"
)

# Also compute unweighted stats over the same -20 dB region
result_unweighted_20db = integrate_delta_psi_db_region(
    delta_psi,
    I_ref=I_avg,
    db_min=-20.0,
    weight=None,
    label="Wire ON - OFF delta psi"
)
```

```python
# ------------------------------------------------------------
# Plot beam-weighted delta-psi contribution with rescaled colorbar
# ------------------------------------------------------------

def plot_beam_weighted_delta_psi(
    x, y,
    delta_psi_deg,
    weight,
    mode="relative",
    title=None,
    cmap="coolwarm",
    percentile=99.5,
    symmetric=True,
):
    """
    Plot a beam-weighted delta-psi map with the colorbar scaled to the
    actual weighted values.

    Parameters
    ----------
    x, y : 1D ndarray
        Coordinate axes.

    delta_psi_deg : 2D ndarray
        Wrapped polarization angle difference map in degrees.

    weight : 2D ndarray
        Beam/intensity weight map, e.g. I_avg = 0.5 * (I_on + I_off).

    mode : str
        "relative":
            plot (weight / max(weight)) * delta_psi_deg.
            Units are still deg, but this shows the local angle contribution
            suppressed by relative beam strength.

        "integral":
            plot (weight / sum(weight)) * delta_psi_deg.
            Units are deg per pixel contribution to the signed weighted mean.
            The sum of this map equals the weighted mean delta psi.

    percentile : float
        Percentile used to set color limits, so one extreme pixel does not
        dominate the colorbar. Use 100 for full min/max scaling.

    symmetric : bool
        If True, use symmetric color limits around zero.
    """
    delta = np.asarray(delta_psi_deg, dtype=float)
    w = np.asarray(weight, dtype=float)

    valid = np.isfinite(delta) & np.isfinite(w) & (w > 0)

    if mode == "relative":
        wnorm = np.zeros_like(w, dtype=float)
        wmax = np.nanmax(w[valid])
        wnorm[valid] = w[valid] / wmax
        plot_map = wnorm * delta
        cbar_label = r"$(I/I_{\max})\,\Delta\psi$ [deg]"
        if title is None:
            title = r"Beam-weighted local angle difference: $(I/I_{\max})\Delta\psi$"

    elif mode == "integral":
        wnorm = np.zeros_like(w, dtype=float)
        wsum = np.nansum(w[valid])
        wnorm[valid] = w[valid] / wsum
        plot_map = wnorm * delta
        cbar_label = r"$(I/\sum I)\,\Delta\psi$ [deg / pixel]"
        if title is None:
            title = r"Per-pixel contribution to weighted mean $\Delta\psi$"

    else:
        raise ValueError("mode must be 'relative' or 'integral'")

    plot_map = np.where(valid, plot_map, np.nan)

    finite_vals = plot_map[np.isfinite(plot_map)]

    if finite_vals.size == 0:
        print("No finite values to plot.")
        return None

    if percentile >= 100:
        vmax = np.nanmax(np.abs(finite_vals)) if symmetric else np.nanmax(finite_vals)
        vmin = -vmax if symmetric else np.nanmin(finite_vals)
    else:
        if symmetric:
            vmax = np.nanpercentile(np.abs(finite_vals), percentile)
            vmin = -vmax
        else:
            vmin = np.nanpercentile(finite_vals, 100 - percentile)
            vmax = np.nanpercentile(finite_vals, percentile)

    fig, ax = plt.subplots(figsize=(6, 5))
    im = ax.imshow(
        plot_map,
        extent=[x.min(), x.max(), y.min(), y.max()],
        origin="lower",
        aspect="equal",
        cmap=cmap,
        vmin=vmin,
        vmax=vmax,
    )

    ax.set_title(title)
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    plt.colorbar(im, ax=ax, label=cbar_label)
    plt.tight_layout()
    plt.show()

    print(f"\n--- {mode} beam-weighted delta psi map ---")
    print(f"min plotted quantity       = {np.nanmin(plot_map):.6e}")
    print(f"max plotted quantity       = {np.nanmax(plot_map):.6e}")
    print(f"max abs plotted quantity   = {np.nanmax(np.abs(plot_map)):.6e}")

    if mode == "integral":
        print(f"sum of plotted map         = {np.nansum(plot_map):.12e} deg")
        print("This should equal the signed beam-weighted mean delta psi.")

    if mode == "relative":
        weighted_mean = np.nansum(w[valid] * delta[valid]) / np.nansum(w[valid])
        print(f"signed beam-weighted mean  = {weighted_mean:.12e} deg")
        print("This plot shows local delta psi suppressed by relative beam intensity.")

    return plot_map


# Use average ON/OFF beam as the weight, same as your summary stats
I_avg = 0.5 * (I_on + I_off)

# Easier-to-read visual version: units are still degrees, scaled by I/Imax
weighted_local_map = plot_beam_weighted_delta_psi(
    x_on, y_on,
    delta_psi,
    weight=I_avg,
    mode="relative",
    percentile=99.5,
    title=r"Beam-weighted $\Delta\psi$: $(I/I_{\max})(\psi_{\rm on}-\psi_{\rm off})$"
)

# Literal contribution-to-integral version: sum of map equals weighted mean
weighted_integral_map = plot_beam_weighted_delta_psi(
    x_on, y_on,
    delta_psi,
    weight=I_avg,
    mode="integral",
    percentile=99.5,
    title=r"Per-pixel contr. to integrated $\Delta\psi$: $(I/\sum I)(\psi_{\rm on}-\psi_{\rm off})$"
)
```
